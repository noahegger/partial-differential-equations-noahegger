# Basic Project: $V(x) = 0$

```python
import pylab
import numpy as np 
import matplotlib.pyplot as plt
import f90nml

plt.style.use('fivethirtyeight')
```

## Probability Density Snapshots

```python
parameters = f90nml.read('zero_potential.namelist')

file_name_1 = parameters['output']['density_file']
file_name_2 = parameters['output']['time_file']


file_name = file_name_1
all_data = np.loadtxt(file_name,skiprows=0)

x = all_data[0]
t0 = all_data[1]
t165 = all_data[34]
t335 = all_data[68]
t5 = all_data[100]

plt.plot(x, t0, label = 't=0')
plt.plot(x, t165, label = 't=1.65')
plt.plot(x, t335, label = 't=3.35')
plt.plot(x, t5, label = 't=5')
plt.legend(loc='upper right')
plt.figtext(.21, .6, '$k = 0$')
plt.figtext(.21, .55, '$\sigma = 0.5$')
plt.figtext(.21, .5, '$x_0 = 0$')
plt.title('Probability Density v.s. Position')
plt.xlabel('Position ')
plt.ylabel('Probability Density')
plt.show()
```

## Normalization

```python
file_name = file_name_2
all_data = np.loadtxt(file_name,skiprows=1)
all_data = all_data.transpose()

t = all_data[0]
position = all_data[1]
norm = all_data[2]
width_numerical = all_data[3]
width_analytic = all_data[4]

plt.plot(t, norm)
plt.figtext(.15, .75, r'$k = 0$')
plt.figtext(.15, .70, r'$\sigma = 0.5$')
plt.figtext(.15, .65, r'$x_0 = 0$')
pylab.ylim([norm[1]-0.1*norm[1],norm[1]+0.1*norm[1]])
plt.title('Normalization v.s. Time')
plt.xlabel('Time ')
plt.ylabel('Normalization')
plt.show()
```

## Position

```python
plt.plot(t,position)
plt.figtext(.15, .6, r'$k = 0$')
plt.figtext(.15, .55, r'$\sigma = 0.5$')
plt.figtext(.15, .5, r'$x_0 = 0$')
plt.title('Position v.s. Time')
plt.xlabel('Time ')
plt.ylabel('Position')
plt.show()
```

## Width

```python
plt.plot(t,width_numerical, label = 'Numerical')
plt.plot(t,width_analytic, label = 'Analytical')
plt.legend(loc='upper left')
plt.figtext(.15, .6, r'$k = 0$')
plt.figtext(.15, .55, r'$\sigma = 0.5$')
plt.figtext(.15, .5, r'$x_0 = 0$')
plt.title('Analytic vs. Numerical Width')
plt.xlabel('Time ')
plt.ylabel('Width')
plt.show()
```

# Advanced Project: $V(x) = \frac{1}{2} kx^2$

## Probability Density Snapshots: Harmonic

```python
parameters_2 = f90nml.read('harmonic_potential.namelist')

file_name_3 = parameters_2['output']['density_file']
file_name_4 = parameters_2['output']['time_file']
file_name_5 = parameters_2['output']['position_file']

# Load density file
file_name = file_name_3
all_data = np.loadtxt(file_name,skiprows=0)

x = all_data[0]
t_0 = all_data[1]
t_165 = all_data[34]
t_335 = all_data[68]
t_5 = all_data[100]

plt.plot(x, t_0, label = 't=0')
plt.plot(x, t_165, label = 't=1.65')
plt.plot(x, t_335, label = 't=3.35')
plt.plot(x, t_5, label = 't=5')
plt.legend(loc='upper right')
plt.figtext(.15, .6, '$k = 0.7$')
plt.figtext(.15, .55, '$\sigma = 1.0$')
plt.figtext(.15, .5, '$x_0 = 2.5$')
plt.title('Probability Density v.s. Position')
plt.xlabel('Position')
plt.ylabel('Probability Density')

plt.legend(loc='upper right')
plt.show()
```

## Normalization: Harmonic

```python
# Load time file
file_name = file_name_4
all_data = np.loadtxt(file_name,skiprows=1)
all_data = all_data.transpose()

t_h = all_data[0]
position_h = all_data[1]
norm_h = all_data[2]
width_numerical_h = all_data[3]
width_analytic_h = all_data[4]

plt.plot(t_h, norm_h)
plt.figtext(.15, .75, r'$k = 0.7$')
plt.figtext(.15, .70, r'$\sigma = 1.0$')
plt.figtext(.15, .65, r'$x_0 = 2.5$')
pylab.ylim([norm[1]-0.1*norm[1],norm[1]+0.1*norm[1]])
plt.title('Normalization v.s. Time')
plt.xlabel('Time ')
plt.ylabel('Normalization')
plt.show()
```

## Position: Harmonic

```python
plt.plot(t_h,position_h)
plt.figtext(.15, .6, r'$k = 0.7$')
plt.figtext(.15, .55, r'$\sigma = 1.0$')
plt.figtext(.15, .5, r'$x_0 = 2.5$')
plt.title('Position v.s. Time')
plt.xlabel('Time ')
plt.ylabel('Position')
plt.show()
```

## Width: Harmonic - Numerical v.s. Analytic

```python
plt.plot(t_h,width_numerical_h, label = 'Numerical')
plt.plot(t_h,width_analytic_h, label = 'Analytic')
plt.legend(loc='upper left')
plt.figtext(.75, .6, '$k = 0.7$')
plt.figtext(.75, .55, '$\sigma = 1.0$')
plt.figtext(.75, .5, '$x_0 = 2.5$')
plt.title('Analytic vs. Numerical Width - Harmonic')
plt.xlabel('Time ')
plt.ylabel('Width')
plt.show()
```

## Numerical v.s. Analytic Position: Harmonic

```python
# Load position file
file_name = file_name_5
all_data = np.loadtxt(file_name,skiprows=1)
all_data = all_data.transpose()

position_analytic_h = all_data[1]

plt.plot(t_h,position_analytic_h, label = 'Analytic')
plt.plot(t_h,position_h, label = 'Numerical')
plt.legend(loc='lower right')
plt.figtext(.55, .7, '$k = 0.7$')
plt.figtext(.55, .65, '$\sigma = 1.0$')
plt.figtext(.55, .6, '$x_0 = 2.5$')
plt.title('Numerical vs. Analytic Position')
plt.xlabel('Time ')
plt.ylabel('Position')
plt.show()
```

### Notice above that we do, in fact, closely resemble the expected cosine function for the position as a function of time.
